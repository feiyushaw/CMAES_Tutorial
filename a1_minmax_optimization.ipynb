# 使用 Evolution Strategy 进行 Min-Max 优化

本专题介绍如何使用 Evolution Strategy（ES）处理 Min-Max（极小极大）优化问题，重点讨论两类方法：

1. **Adversarial-CMA-ES** [1]：面向局部鞍点搜索，通过交替求解内层最大化与外层最小化，并用学习率控制两者耦合。参考实现：https://gist.github.com/youheiakimoto/ab51e88c73baf68effd95b750100aad0
2. **WRA-CMA-ES** [2]：面向更全局的搜索和并行计算，不要求精确得到每个候选解的最坏性能，而是近似它们的最坏情况排名。参考实现：https://github.com/akimotolab/worstcase-ranking-approximation

[1] Youhei Akimoto, Yoshiki Miyauchi, and Atsuo Maki. *Saddle Point Optimization with Approximate Minimization Oracle and Its Application to Robust Berthing Control*. ACM TELO, 2022. https://doi.org/10.1145/3510425

[2] Atsuhiro Miyagi, Yoshiki Miyauchi, Atsuo Maki, Kazuto Fukuchi, Jun Sakuma, and Youhei Akimoto. *Covariance Matrix Adaptation Evolutionary Strategy with Worst-Case Ranking Approximation for Min-Max Optimization and Its Application to Berthing Control Tasks*. ACM TELO, 2023. https://doi.org/10.1145/3603716

## 背景：仿真参数与真实运行条件的不确定性

ES 经常用于数值仿真驱动的黑盒优化。给定设计变量 $x$，通过仿真得到目标函数 $f(x)$，然后搜索更优的 $x$。实际问题中，仿真结果通常还依赖一些无法在设计阶段准确确定的条件，例如路面状态、天气、载荷、环境扰动或模型参数。下面统一把这类不确定因素记为 $y$，于是目标函数写成 $f(x,y)$。

如果在优化阶段固定一个仿真条件 $y_{\mathrm{sim}}$，实际上求解的是
$$x^*=\operatorname{argmin}_x f(x,y_{\mathrm{sim}}).$$
但部署时遇到的真实条件 $y$ 往往不同于 $y_{\mathrm{sim}}$，因此在仿真条件下表现很好的 $x^*$，在真实条件下可能明显退化。鲁棒优化的目标就是避免这种对单一场景的过拟合。

### 两个简单例子

#### 1. 凸-凹二次函数
考虑
$$f(x,y)=a x^2+bxy-cy^2,\qquad a,c>0.$$
对固定的 $y$，最优 $x$ 为
$$x^*(y)=-\frac{by}{2a}.$$
因此 $x$ 的最优值随 $y$ 改变，且 $|b|/a$ 越大，二者耦合越强。固定 $y_{\mathrm{sim}}$ 求出的设计用于另一条件 $y$ 时，会多出与 $(y_{\mathrm{sim}}-y)^2$ 成比例的性能损失。

#### 2. 双线性函数
考虑定义域 $[-1,1]\times[-1,1]$ 上的
$$f(x,y)=xy.$$
对给定 $y$，最优解为 $x^*(y)=-\operatorname{sign}(y)$。当 $y$ 的符号变化时，最优 $x$ 会发生不连续跳变。这说明不确定参数不仅可能平滑地改变最优设计，也可能让最优解发生结构性变化。

## Min-Max 优化问题

假设部署阶段可能出现的不确定条件组成集合 $Y$。对一个给定的设计 $x$，定义其最坏性能
$$F_Y(x)=\max_{y\in Y}f(x,y).$$
鲁棒设计的目标是最小化这个最坏性能：
$$x^*=\operatorname{argmin}_x F_Y(x)=\operatorname{argmin}_x\max_{y\in Y}f(x,y).$$

在黑盒设置中，我们通常不能解析地得到 $F_Y(x)$，因为每计算一次 $f(x,y)$ 都可能需要昂贵仿真。因此真正的困难并不只是外层优化 $x$，还包括如何高效估计每个 $x$ 对应的最坏场景。

## 朴素方法：有限场景采样

最直接的方法是预先采样 $M$ 个场景 $y_1,\dots,y_M$，然后用
$$\tilde F_Y(x)=\max_{m=1,\dots,M}f(x,y_m)$$
近似真实最坏性能 $F_Y(x)$。如果 $M$ 个场景可以并行仿真，这种方法工程上非常简单，也可以直接把 $\tilde F_Y$ 交给普通 CMA-ES。

它的核心缺陷是：有限样本只覆盖了 $Y$ 的很小一部分。尤其当 $y$ 是高维变量时，即使 $M$ 看起来不小，也可能遗漏真正危险的区域。因为 $\tilde F_Y(x)\le F_Y(x)$，这种方法会系统性低估最坏性能。

```python
import numpy as np
import matplotlib.pyplot as plt

a = c = 1.0
b = 10.0

def f(x, y):
    return a * np.dot(x, x) + b * np.dot(x, y) - c * np.dot(y, y)

def f_worst(x):
    # 对该二次函数，内层最大化有解析解
    y = b * x / (2 * c)
    return f(x, y)

np.random.seed(100)
M = 10
y_samples = np.random.randn(M, 1)

def f_worst_approx(x):
    return max(f(x, y) for y in y_samples)
```

```python
x_array = np.linspace(-1, 1, 300)
true_values = np.array([f_worst(np.array([x])) for x in x_array])
approx_values = np.array([f_worst_approx(np.array([x])) for x in x_array])
plt.plot(x_array, true_values, label=r'$F_Y$')
plt.plot(x_array, approx_values, label=r'$\tilde F_Y$')
plt.grid()
plt.legend()
```

在高维不确定性空间中，有限场景方法通常面临两个问题：

1. **覆盖不足**：例如 $y$ 为 10 维而只采样 10 个场景时，几乎不可能充分逼近真实最坏情况。
2. **非光滑性**：$\tilde F_Y(x)$ 是多个函数 $f(x,y_m)$ 的最大值，它的局部最优点常位于不同场景曲面的交界处，因此即使每个 $f(x,y_m)$ 都很光滑，$\tilde F_Y$ 也可能存在折角。

第一个问题通常更根本：如果最危险场景根本没有被采到，再精确优化 $\tilde F_Y$ 也无法得到真正鲁棒的解。

## Adversarial-CMA-ES

Adversarial-CMA-ES 的目标是寻找 $f(x,y)$ 的 Min-Max 鞍点。局部 Min-Max 鞍点 $(x^*,y^*)$ 满足：在其邻域内，$x^*$ 对固定 $y^*$ 是局部最小，而 $y^*$ 对固定 $x^*$ 是局部最大。直观地说，$x$ 是设计者，$y$ 是对抗环境；算法让两者交替寻找各自最优响应。

如果问题是严格凸-凹的，Min-Max 最优解与鞍点一致；一般非凸非凹问题则没有这个保证，因此找到局部鞍点并不自动意味着找到了全局鲁棒最优解。

### 为什么简单交替优化会发散

仍考虑
$$f(x,y)=ax^2+bxy-cy^2.$$
精确的最优响应为
$$x^*(y)=-\frac{by}{2a},\qquad y^*(x)=\frac{bx}{2c}.$$
如果机械地交替执行 $y^{(t)}=y^*(x^{(t)})$、$x^{(t+1)}=x^*(y^{(t)})$，则
$$x^{(t)}=\left(-\frac{b^2}{4ac}\right)^t x^{(0)}.$$
当 $b^2>4|ac|$ 时，这个序列会发散。也就是说，$x$ 与 $y$ 的耦合越强，单纯的“你一步、我一步”越容易发生过冲和振荡。

### Adversarial-CMA-ES 的关键：阻尼更新

不直接跳到对方当前条件下的最优响应，而是只沿该方向移动一小步：
$$\begin{aligned}
x^{(t+1)} &= x^{(t)}+\eta(x^*(y^{(t)})-x^{(t)}),\\
y^{(t+1)} &= y^{(t)}+\eta(y^*(x^{(t)})-y^{(t)}).
\end{aligned}$$
其中 $\eta>0$ 是学习率。令 $a=c=1$，可以写成
$$z^{(t+1)}=Az^{(t)},\qquad A=\begin{bmatrix}1-\eta&-\frac b2\eta\\\frac b2\eta&1-\eta\end{bmatrix}.$$
矩阵 $A$ 的两个奇异值均为
$$\sqrt{(1-\eta)^2+\frac{b^2}{4}\eta^2}.$$
当
$$\eta<\frac{8}{4+b^2}$$
时奇异值小于 1，因此迭代收敛。这里也清楚显示了一个重要限制：耦合参数 $b$ 越大，可接受学习率越小，因此即使算法保持稳定，收敛也会变慢。

实际黑盒问题中无法精确计算 $x^*(y)$ 与 $y^*(x)$。Adversarial-CMA-ES 使用 (1+1)-CMA-ES 分别近似内层最优化，并允许这些近似带有误差；只要估计的更新方向仍基本正确，小学习率可以吸收一部分内层求解误差。

### Adversarial-CMA-ES 的实现要点

* $x$ 与 $y$ 的局部最优响应都由 (1+1)-CMA-ES 近似求解。理论上，在光滑强凸问题中 (1+1)-CMA-ES 具有线性收敛，因此每次内层调用不需要追求极高精度。
* 学习率 $\eta$ 不预先固定，而是利用近似 gap function
$$G(x,y)=\max_{y'}f(x,y')-\min_{x'}f(x',y)$$
的下降速度自适应调整。鞍点处该 gap 应趋近于 0。
* 对一般非凸非凹问题，局部 Min-Max 鞍点可能不是最终希望的鲁棒解，因此算法会结合重启，并保存历次发现的困难场景 $y_k$，让后续搜索同时对这些场景保持鲁棒。

完整参考实现见：https://gist.github.com/youheiakimoto/ab51e88c73baf68effd95b750100aad0

## WRA-CMA-ES

WRA-CMA-ES（Worst-case Ranking Approximation CMA-ES）不以寻找鞍点为核心，而是直接把
$$F_Y(x)=\max_{y\in Y}f(x,y)$$
作为外层 CMA-ES 的目标。它主要针对 Adversarial-CMA-ES 的两个局限：

* 当问题不是局部强凸-强凹时，鞍点式交替优化可能不收敛到 Min-Max 最优。
* 当 $x$ 与 $y$ 强耦合时，Adversarial-CMA-ES 必须使用很小的学习率，速度会明显下降。

WRA 不需要精确计算每个 $F_Y(x_i)$。CMA-ES 更新实际上只依赖候选解之间的**排序**，因此只要能够可靠判断 $F_Y(x_1),\dots,F_Y(x_\lambda)$ 的相对次序，就足以完成外层更新。

### Worst-case Ranking Approximation 的两个关键策略

#### 1. Warm Start
外层 CMA-ES 相邻两代的分布通常变化不会非常剧烈。如果最坏场景随 $x$ 的变化较为连续，那么上一代求得的最坏场景和内层 CMA-ES 状态，就是下一代非常好的初始化。因此 WRA 会跨外层迭代保留内层优化器的状态，而不是每次从零开始。

此外，WRA 同时维护多个内层 CMA-ES 实例。这样即使不同 $x$ 对应的最坏场景分属于不同模式，也能同时保留多条“危险场景轨迹”，避免单个内层优化器在模式切换时反复重新搜索。

#### 2. Early Stop
内层最大化没有必要全部收敛到很高精度。WRA 会同步推进各候选解的最坏场景搜索，并周期性计算当前估计的最坏性能排名。若连续两次排名之间的 Kendall 秩相关已经足够接近 1，就说明候选解排序基本稳定，此时即可停止内层求解并更新外层 CMA-ES。

因此 WRA 的核心思想是：**为排序服务，而不是为每个最坏函数值求高精度数值。**

### 重启与局部精修

如果 $F_Y(x)$ 本身是多峰函数，一次 WRA-CMA-ES 仍可能落入局部解。因此算法可以在满足单次终止条件后保存当前候选设计集合 $X_{hist}$ 与场景集合 $Y_{hist}$，重新初始化并继续搜索。最终从历史集合中选择
$$\min_{x\in X_{hist}}\max_{y\in Y_{hist}}f(x,y)$$
对应的设计。

论文还提出可选的混合策略：先用 WRA-CMA-ES 做全局搜索，再用 Adversarial-CMA-ES 在当前解附近做局部精修。对于局部近似强凸-强凹的问题，后者往往能更快获得高精度。

完整 WRA-CMA-ES 实现与论文实验代码见：https://github.com/akimotolab/worstcase-ranking-approximation

## Adversarial-CMA-ES 与 WRA-CMA-ES 如何选择

在凸-凹二次问题中，可以通过改变交互系数 $b$ 直观比较两者：

* 当 $b$ 较小、局部强凸-强凹结构明显时，Adversarial-CMA-ES 往往收敛更快，因为它直接利用局部 Min-Max 结构。
* 当 $b$ 变大时，Adversarial-CMA-ES 为保持稳定必须降低 $\eta$，收敛会显著变慢。WRA 不采用这种交替阻尼更新，因此对耦合强度通常不那么敏感。
* 对双线性函数 $f(x,y)=x^Ty$，简单鞍点式交替机制可能持续旋转或振荡，而 WRA 直接优化最坏性能排序，能够规避这一结构性问题。

可以把两者的适用范围概括为：**Adversarial-CMA-ES 更像高效局部 Min-Max 求解器；WRA-CMA-ES 更像面向一般黑盒鲁棒优化的外层全局搜索器。**

```python
# 一个用于理解交互强度的简单实验：阻尼交替最优响应
def damped_best_response(b, eta, x0=1.0, y0=1.0, steps=100):
    x, y = x0, y0
    hist = []
    for _ in range(steps):
        x_best = -b * y / 2.0
        y_best =  b * x / 2.0
        x_new = x + eta * (x_best - x)
        y_new = y + eta * (y_best - y)
        x, y = x_new, y_new
        hist.append((x, y))
    return np.asarray(hist)

for b in [2.0, 8.0]:
    eta = 0.5 * 8.0 / (4.0 + b**2)
    h = damped_best_response(b=b, eta=eta)
    plt.semilogy(np.linalg.norm(h, axis=1), label=f'b={b}, eta={eta:.3g}')
plt.xlabel('iteration')
plt.ylabel(r'$\|(x,y)\|$')
plt.grid()
plt.legend()
```

## 进一步学习建议

WRA-CMA-ES 仓库中包含论文使用的多种 Min-Max 测试函数。建议不要只比较最终函数值，还应观察以下现象：

* 最坏场景是否随 $x$ 平滑变化，还是会发生模式切换；
* 外层候选解排名需要多少次内层函数评估才能稳定；
* 强耦合、非凸非凹、多峰内层最大化分别如何影响两种算法；
* Warm Start、多个内层优化器以及 Early Stop 分别节省了多少函数评估。

这些实验比只记住算法名称更有助于理解 Min-Max 黑盒优化真正困难的部分。
